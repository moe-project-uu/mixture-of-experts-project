```python
import numpy as np
import torch
```

```python
B = 5
D = 20
def expert(x):
    print(x)
experts = [expert, expert, expert, expert]
h = torch.randn(B, D) #this is a matrix of our input data from the latent space h
print(h)
```

```python
C=10
logits = h.new_zeros(B, C)
print(logits)
```

```python
topk_idx=torch.tensor([[0, 1],
                      [1, 3],
                      [2, 3],
                      [1, 2],
                      [1, 2]])
print(topk_idx)
```

```python
for e, expert in enumerate(experts): 
    mask = (topk_idx == e).any(dim=1) #for each expert index 0, 1, 2, 3, we check if that expert index is in the topk for each of the 5 images
    print(mask)
    h_e = h[mask]
    print(f"{e}: {h_e}") #for each expert e, h_e is a tensor of all input data for which that expert is in the topk
    #dim of h_e is (B_e, D) where B_e is the number of datapoints where the expert is in the topk for 
```
